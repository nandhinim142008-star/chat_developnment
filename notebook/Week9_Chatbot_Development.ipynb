```python
import pandas as pd
from pathlib import Path

BASE_DIR = Path.cwd().parent
QUESTIONS_FILE = BASE_DIR / "dataset" / "sample_questions.csv"

questions_df = pd.read_csv(QUESTIONS_FILE)

print("Total Sample Questions:", len(questions_df))
display(questions_df)
```

```python
# Count questions by question type
question_summary = questions_df["QUESTION_TYPE"].value_counts()

print("Question Type Summary:")
display(question_summary)

print("\nQuestion Dataset Information:")
print("Total Questions:", len(questions_df))
print("Total Columns:", len(questions_df.columns))
print("Column Names:", list(questions_df.columns))
print("Missing Values:", questions_df.isnull().sum().sum())
```

```python
import ollama

MODEL_NAME = "gemma3:1b"

try:
    response = ollama.chat(
        model=MODEL_NAME,
        messages=[
            {
                "role": "user",
                "content": "Explain artificial intelligence in two simple sentences."
            }
        ]
    )

    print("Ollama connection successful!")
    print("Model:", MODEL_NAME)
    print("\nChatbot Response:")
    print(response["message"]["content"])

except Exception as error:
    print("Connection Error:", error)
```

```python

import ollama

MODEL_NAME = "gemma3:1b"

messages = [
    {
        "role": "system",
        "content": "You are a friendly education chatbot. Answer in simple English."
    }
]

# First question
messages.append({
    "role": "user",
    "content": "My name is Ravi. I am learning Python."
})

response1 = ollama.chat(model=MODEL_NAME, messages=messages)
answer1 = response1["message"]["content"]
messages.append({"role": "assistant", "content": answer1})

print("Question 1: My name is Ravi. I am learning Python.")
print("Chatbot:", answer1)

# Follow-up question using previous context
messages.append({
    "role": "user",
    "content": "What programming language am I learning?"
})

response2 = ollama.chat(model=MODEL_NAME, messages=messages)
answer2 = response2["message"]["content"]

print("\nQuestion 2: What programming language am I learning?")
print("Chatbot:", answer2)
```

```python

import ollama
import pandas as pd
from pathlib import Path
from datetime import datetime

MODEL_NAME = "gemma3:1b"
BASE_DIR = Path.cwd().parent
DATASET_DIR = BASE_DIR / "dataset"

responses = []

for index, row in questions_df.iterrows():
    question = row["QUESTION"]
    print(f"Processing {index + 1}/20: {question}")

    try:
        result = ollama.chat(
            model=MODEL_NAME,
            messages=[
                {
                    "role": "system",
                    "content": (
                        "You are a friendly education chatbot. "
                        "Answer clearly in simple English."
                    )
                },
                {"role": "user", "content": question}
            ]
        )

        answer = result["message"]["content"]

        responses.append({
            "TIMESTAMP": datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
            "QUESTION_ID": row["QUESTION_ID"],
            "QUESTION": question,
            "RESPONSE": answer,
            "MODEL": MODEL_NAME
        })

    except Exception as error:
        print("Error:", error)

responses_df = pd.DataFrame(responses)
responses_df.to_csv(DATASET_DIR / "responses.csv", index=False)

print("\nResponses saved:", len(responses_df))
print("File:", DATASET_DIR / "responses.csv")
```
